## Exercise 3: grade(score): if / elif / else  `easy`

Return `"A"` for 90+, `"B"` for 80-89, `"C"` for 70-79, `"D"` for 60-69, otherwise `"F"`.
Raise `ValueError` if the score is below 0 or above 100.

````python
def grade(score):
    """Return the letter grade for a 0-100 score."""
    if score <0 or score >100:
        raise ValueError
    if score >= 90:
        return "A"
    elif score >=80 and score <90:
        return "B"
    elif score >=70 and score <80:
        return "C"
    elif score >=60 and score <=70:
        return "D"
    else:
        return "F"
````

````python
def _test():
    assert grade(95) == "A" and grade(90) == "A", "90 and above is an A"
    assert grade(89.9) == "B", "89.9 is a B"
    assert grade(70) == "C" and grade(60) == "D" and grade(59) == "F" and grade(0) == "F"
    assert raises(ValueError, grade, 101), "101 should raise ValueError"
    assert raises(ValueError, grade, -1), "-1 should raise ValueError"

check('3. grade(score): if / elif / else', _test)
````

## Exercise 4: pay(hours, rate): the classic py4e exercise  `easy`

Pay is `hours * rate`, but hours above 40 are paid at **1.5x** the rate. Return a `float`.

`pay(45, 10)` is `40*10 + 5*15 = 475.0`.

````python
def pay(hours, rate):
    """Pay with time-and-a-half for hours over 40."""
    total : float = 0
    for ind,hr in enumerate(range(hours)):
        if ind+1>40:
            total = total + rate+(rate/2)
        else:
            total = total + rate
    return float(total)

````

````python
def _test():
    assert pay(45, 10) == 475.0, "45 hours at 10: 400 + 5 * 15"
    assert pay(40, 10) == 400.0, "exactly 40 hours has no overtime"
    assert pay(10, 10) == 100.0
    assert isinstance(pay(10, 10), float), "return a float"

check('4. pay(hours, rate): the classic py4e exercise', _test)
````

## Exercise 5: average_until_done(entries): loops, break, continue  `medium`

`entries` is a list of strings, like what a user would type. Go through them in order:
- stop at the first `"done"` (`break`)
- skip anything that isn't a number (`continue`, use `try/except ValueError` around `float()`)

Return `(count, total, average)`. If there were no valid numbers, the average is `None`.

<details><summary>Hint</summary>

`float('x')` raises `ValueError`.

</details>

````python
def average_until_done(entries):
    """Return (count, total, average) of the numbers before \"done\"."""
    count, total = 0, 0.0
    for item in entries:
        if item == "done":
            break
        try:
            value = float(item)
        except ValueError:
            continue
        count+=1
        total = value + total
    avg = total/count if count else None
    return (count, total, avg)
````

````python
def _test():
    assert average_until_done(["4", "6", "x", "8", "done", "100"]) == (3, 18.0, 6.0), "the 100 comes after done, and x is skipped"
    assert average_until_done([]) == (0, 0, None)
    assert average_until_done(["done", "5"]) == (0, 0, None)
    assert average_until_done(["2.5", "abc", "3.5"]) == (2, 6.0, 3.0)

check('5. average_until_done(entries): loops, break, continue', _test)
````

## Exercise 6: summarize(*numbers, round_to=2): *args and keyword-only  `medium`

Accept any number of numbers, plus a **keyword-only** `round_to` (default 2). Return a dict `{"count", "total", "mean"}`. The mean is rounded to `round_to` places, or `None` when there are no numbers.

`summarize(1, 2, 4)` -> `{"count": 3, "total": 7, "mean": 2.33}`

<details><summary>Hint</summary>

Put `*numbers` first, then `round_to=2`. Anything after `*numbers` can only be passed by name.

</details>

````python
def summarize(*numbers, round_to=2):
    count = 0
    total = 0
    mean = 0
    for num in numbers:
        print(num)
        count += 1
        total = total + num
    mean = round(total/count, round_to) if count else None
    return {
        'count':count,
        'total':total,
        'mean': mean
    }
````

````python
def _test():
    assert summarize(1, 2, 4) == {"count": 3, "total": 7, "mean": 2.33}, "expected count 3, total 7, mean 2.33"
    assert summarize() == {"count": 0, "total": 0, "mean": None}, "no numbers means mean is None"
    assert summarize(1, 2, 4, round_to=0)["mean"] == 2
    assert summarize(10)["mean"] == 10

check('6. summarize(*numbers, round_to=2): *args and keyword-only', _test)
````

## Exercise 7: Interview: fix the mutable default argument  `interview`

This function has the most famous Python bug:

```python
def add_tag(tag, tags=[]):
    tags.append(tag)
    return tags

add_tag("a")   # ['a']
add_tag("b")   # ['a', 'b']   <- surprise!
```

Write `add_tag_fixed(tag, tags=None)`. Each call without `tags` must start with a fresh list. If a list **is** passed, append to that same list and return it.

Then explain *why* the original is buggy (it's in the Say-it questions below).

<details><summary>Hint</summary>

Default values are evaluated **once**, when the function is defined. Use `None` as the default and create the list inside.

</details>

````python
def add_tag_fixed(tag, tags=None):
    if tags is None:
        tags = []
    tags.append(tag)
    return tags
````

````python
def _test():
    first = add_tag_fixed("a")
    second = add_tag_fixed("b")
    assert first == ["a"], "first call should return ['a']"
    assert second == ["b"], "second call must NOT remember 'a'"
    mine = ["x"]
    result = add_tag_fixed("y", mine)
    assert result is mine and mine == ["x", "y"], "when a list is passed, append to that same list"

check('7. Interview: fix the mutable default argument', _test)
````
