## Exercise 2: word_frequency(text): count with a dict  `easy`

Return a dict of how often each word appears. Lower-case everything and remove punctuation first.

`word_frequency("The cat, the hat. THE end!")` -> `{"the": 3, "cat": 1, "hat": 1, "end": 1}`

<details><summary>Hint</summary>

`str.maketrans('', '', string.punctuation)` + `.translate(...)` removes punctuation. Count with `d[w] = d.get(w, 0) + 1`.

</details>

````python
import string


def word_frequency(text):
    cleaned = text.split()
    my_dict : dict = {}
    for word in cleaned:
        cleaned_word = "".join(c.lower() for c in word if c.isalnum())
        print(cleaned_word)
        my_dict[cleaned_word] = my_dict.get(cleaned_word, 0) + 1
    # print('my',my_dict)
    return my_dict
````

````python
def _test():
    assert word_frequency("The cat, the hat. THE end!") == {"the": 3, "cat": 1, "hat": 1, "end": 1}
    assert word_frequency("") == {}

check('2. word_frequency(text): count with a dict', _test)
````

## Exercise 3: top_n(freq, n): sort a dict by value  `medium`

Given a frequency dict, return the `n` most common as a list of `(word, count)` tuples, highest count first. Break ties alphabetically.

`top_n({"b": 2, "a": 2, "c": 5, "d": 1}, 2)` -> `[("c", 5), ("a", 2)]`

<details><summary>Hint</summary>

`sorted(freq.items(), key=lambda kv: (-kv[1], kv[0]))[:n]`

</details>

````python
def top_n(freq, n):
    return sorted(freq.items(), key=lambda item: (-item[1],item[0]))[0:n]
````

````python
def _test():
    freq = {"b": 2, "a": 2, "c": 5, "d": 1}
    assert top_n(freq, 2) == [("c", 5), ("a", 2)], "highest count first, ties alphabetical"
    assert top_n(freq, 10) == [("c", 5), ("a", 2), ("b", 2), ("d", 1)], "n larger than the dict returns everything"
    assert top_n({}, 3) == []

check('3. top_n(freq, n): sort a dict by value', _test)
````

## Exercise 4: invert(d): values become keys  `medium`

Invert a dict. Several keys can share a value, so each new value maps to a **sorted list** of the original keys.

`invert({"a": 1, "b": 2, "c": 1})` -> `{1: ["a", "c"], 2: ["b"]}`

<details><summary>Hint</summary>

`out.setdefault(value, []).append(key)`

</details>

````python
def invert(d):
    my_dict = {}
    for item in d.items():
        if not my_dict.get(item[1]):
            my_dict[item[1]] = [item[0]]
        else:
            my_dict[item[1]].append(item[0])
    return my_dict
````

````python
def _test():
    assert invert({"a": 1, "b": 2, "c": 1}) == {1: ["a", "c"], 2: ["b"]}
    assert invert({}) == {}

check('4. invert(d): values become keys', _test)
````

## Exercise 6: compare_users(a, b): set operations  `medium`

Two lists of usernames may contain duplicates. Return a dict with three **sorted lists**: `"both"`, `"only_a"`, `"only_b"`.

<details><summary>Hint</summary>

`set(a) & set(b)`, `set(a) - set(b)`, `set(b) - set(a)`

</details>

````python
def compare_users(a, b):
    set_a, set_b = set(a), set(b)
    return {
        "both": sorted(set_a & set_b),
        "only_a":sorted(set_a - set_b),
        "only_b":sorted(set_b-set_a)
    }
````

````python
def _test():
    result = compare_users(["x", "y", "y", "z"], ["y", "z", "w"])
    assert result == {"both": ["y", "z"], "only_a": ["x"], "only_b": ["w"]}, "use sets and sort each result"

check('6. compare_users(a, b): set operations', _test)
````

## Exercise 7: Interview: two_sum(nums, target)  `interview`

Return a tuple `(i, j)` with `i < j` of two **different** positions whose values add up to `target`, or `None` if there is no such pair. Aim for one pass (`O(n)`) using a dict instead of two nested loops.

`two_sum([2, 7, 11, 15], 9)` -> `(0, 1)`

<details><summary>Hint</summary>

While looping, remember each number's index in a dict. For each `n`, check whether `target - n` was already seen.

</details>

````python
def two_sum(nums, target):
    my_dict = {}
    for idx,item in enumerate(nums):
        to_find = target-item
        if to_find in my_dict:
            return (my_dict[to_find],idx)
        else:
            my_dict[item]=idx
    return None
````

````python
def _test():
    assert two_sum([2, 7, 11, 15], 9) == (0, 1)
    assert two_sum([3, 2, 4], 6) == (1, 2)
    assert two_sum([3, 3], 6) == (0, 1), "duplicates are allowed, but not the same position twice"
    assert two_sum([1, 2], 10) is None
    assert two_sum([5], 10) is None, "one number can't pair with itself"

check('7. Interview: two_sum(nums, target)', _test)
````

## Exercise 8: Tuples: min_max and swap_pairs  `easy`

1. `min_max(nums)` returns the tuple `(smallest, largest)`.
2. `swap_pairs(pairs)` takes a list of 2-tuples and returns a new list with each pair reversed: `[(1, "a")]` -> `[("a", 1)]`. Use tuple unpacking in a comprehension.

<details><summary>Hint</summary>

`[(b, a) for a, b in pairs]`

</details>

````python
def min_max(nums):
    return (min(nums),max(nums))


def swap_pairs(pairs):
    return [item[::-1] for item in pairs]
````

````python
def _test():
    lo, hi = min_max([4, 1, 9])
    assert (lo, hi) == (1, 9), "min_max should return (1, 9)"
    assert min_max([5]) == (5, 5)
    assert swap_pairs([(1, "a"), (2, "b")]) == [("a", 1), ("b", 2)]
    assert swap_pairs([]) == []

check('8. Tuples: min_max and swap_pairs', _test)
````
