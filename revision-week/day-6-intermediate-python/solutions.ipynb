## Exercise 1: Comprehensions: squares_of_evens and index_by_id  `warm-up`

1. `squares_of_evens(nums)`: a list of `n * n` for the even numbers only, using a list comprehension.
2. `index_by_id(records)`: `records` is a list of dicts with an `"id"` key. Return `{id: record}` using a dict comprehension.

```python
def squares_of_evens(nums):
    return [n * n for n in nums if n % 2 == 0]


def index_by_id(records):
    return {r["id"]: r for r in records}
```

```python
def _test():
    assert squares_of_evens([1, 2, 3, 4, 5, 6]) == [4, 16, 36]
    assert squares_of_evens([]) == []
    records = [{"id": 7, "name": "a"}, {"id": 9, "name": "b"}]
    assert index_by_id(records) == {7: {"id": 7, "name": "a"}, 9: {"id": 9, "name": "b"}}

check('1. Comprehensions: squares_of_evens and index_by_id', _test)
```

## Exercise 2: sort_people(people): sorted with a key  `easy`

Return a **new** list sorted by age, **oldest first**, and by name A-Z when ages are equal. Don't modify the original list.

```python
def sort_people(people):
    return sorted(people, key=lambda p: (-p["age"], p["name"]))
```

```python
def _test():
    people = [{"name": "B", "age": 30}, {"name": "A", "age": 30}, {"name": "C", "age": 40}]
    original = list(people)
    result = sort_people(people)
    assert [p["name"] for p in result] == ["C", "A", "B"], "oldest first, ties by name"
    assert people == original, "the input list must not be modified"

check('2. sort_people(people): sorted with a key', _test)
```

## Exercise 3: batched(iterable, size): a generator  `medium`

Write a **generator function** that yields lists of `size` items at a time. The last list may be shorter. It must work lazily, even on an infinite iterator.

`list(batched(range(7), 3))` -> `[[0, 1, 2], [3, 4, 5], [6]]`

This is how you process a big file or API in chunks without loading it all.

```python
def batched(iterable, size):
    batch = []
    for item in iterable:
        batch.append(item)
        if len(batch) == size:
            yield batch
            batch = []
    if batch:
        yield batch
```

```python
def _test():
    assert inspect.isgeneratorfunction(batched), "use yield so this is a generator function"
    assert list(batched(range(7), 3)) == [[0, 1, 2], [3, 4, 5], [6]]
    assert list(batched([], 3)) == []
    assert next(batched(itertools.count(), 3)) == [0, 1, 2], "must be lazy: it should work on an infinite iterator"

check('3. batched(iterable, size): a generator', _test)
```

## Exercise 4: fib() generator and first_fibs(n)  `medium`

1. `fib()` is an **infinite** generator: `0, 1, 1, 2, 3, 5, ...`
2. `first_fibs(n)` returns the first `n` values as a list, using `itertools.islice` on `fib()`.

```python
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b


def first_fibs(n):
    return list(itertools.islice(fib(), n))
```

```python
def _test():
    assert first_fibs(8) == [0, 1, 1, 2, 3, 5, 8, 13]
    assert first_fibs(0) == []
    g = fib()
    assert (next(g), next(g), next(g)) == (0, 1, 1), "fib() should produce values on demand with next()"

check('4. fib() generator and first_fibs(n)', _test)
```

## Exercise 5: count_calls: your first decorator  `medium`

Write a decorator `count_calls` that counts how many times the wrapped function was called, exposed as `func.calls`. Use `functools.wraps` so the function keeps its `__name__` and `__doc__`.

```python
def count_calls(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper
```

```python
def _test():
    @count_calls
    def greet(name):
        """Say hi."""
        return f"hi {name}"

    assert greet("a") == "hi a", "the decorated function must still return its result"
    greet("b")
    assert greet.calls == 2, "calls should be 2 after two calls"
    assert greet.__name__ == "greet" and greet.__doc__ == "Say hi.", "use functools.wraps to keep the name and docstring"

check('5. count_calls: your first decorator', _test)
```

## Exercise 6: retry(times): a decorator with arguments  `interview`

Write `retry(times)` so `@retry(3)` calls the function up to 3 times. If a call raises, try again. When every attempt fails, raise the **last** exception. If one succeeds, return its result.

This three-level structure (decorator factory, decorator, wrapper) is a classic interview question, and very real for flaky API calls.

```python
def retry(times):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            last = None
            for _ in range(times):
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    last = e
            raise last

        return wrapper

    return decorator
```

```python
def _test():
    calls = []

    @retry(3)
    def flaky():
        calls.append(1)
        if len(calls) < 3:
            raise ConnectionError("not yet")
        return "ok"

    assert flaky() == "ok", "it should succeed on the third attempt"
    assert len(calls) == 3

    attempts = []

    @retry(2)
    def always_fails():
        attempts.append(1)
        raise ValueError("nope")

    assert raises(ValueError, always_fails), "after 2 failed attempts the last exception should be raised"
    assert len(attempts) == 2

check('6. retry(times): a decorator with arguments', _test)
```

## Exercise 7: fib_memo(n): functools.cache  `medium`

Write the plain recursive Fibonacci (`fib_memo(0) == 0`, `fib_memo(1) == 1`) and make it fast with the `functools.cache` decorator. Without it, `fib_memo(80)` would take longer than your lifetime.

```python
@functools.cache
def fib_memo(n):
    if n < 2:
        return n
    return fib_memo(n - 1) + fib_memo(n - 2)
```

```python
def _test():
    assert fib_memo(10) == 55
    assert fib_memo(80) == 23416728348467685, "this needs caching to finish quickly"

check('7. fib_memo(n): functools.cache', _test)
```

## Exercise 8: Interview: anagram_groups(words) with defaultdict  `interview`

Group words that are anagrams of each other. Return a list of groups; each group is sorted A-Z, and the groups are sorted too.

`anagram_groups(["eat", "tea", "tan", "ate", "nat", "bat"])` -> `[["ate", "eat", "tea"], ["bat"], ["nat", "tan"]]`

```python
def anagram_groups(words):
    groups = defaultdict(list)
    for word in words:
        groups["".join(sorted(word))].append(word)
    return sorted(sorted(group) for group in groups.values())
```

```python
def _test():
    words = ["eat", "tea", "tan", "ate", "nat", "bat"]
    assert anagram_groups(words) == [["ate", "eat", "tea"], ["bat"], ["nat", "tan"]]
    assert anagram_groups([]) == []

check('8. Interview: anagram_groups(words) with defaultdict', _test)
```

**Generator vs list: what's the difference and when would you choose a generator?**

A list holds all items in memory. A generator produces one item at a time on demand and holds almost nothing. Use generators for big or infinite streams (files, API pages) and pipelines; use a list when you need random access or several passes.

**What is a decorator?**

A function that takes a function and returns a new function that wraps it, adding behaviour (logging, timing, retries, caching, auth) without changing the original. `@d` above `def f` is shorthand for `f = d(f)`. Use `functools.wraps` to keep metadata.

**What is a closure?**

An inner function that remembers variables from the enclosing function's scope even after the outer function has returned. Decorators and factories rely on this.

**`lambda` vs `def`?**

A lambda is an anonymous function limited to a single expression, handy for short `key=` arguments. Use `def` for anything non-trivial or reused, so it has a name and docstring.

**How do you make a slow recursive function faster?**

Memoization: cache results of previous calls (`functools.cache` / `lru_cache`) so each distinct input is computed once. For Fibonacci this turns exponential time into linear.
